```python
#printing 1 to 10 numbers 10 times
for i in range(1,11):
 for j in range(1,11):
  print(j,end=" ")
 print("\n")
```

```python
#sum of digits
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```

```python
#print prime numbers between given 2 no's
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
   if(i%j==0):
    break
  else:
   print(i)
```

```python
#prime or not
n=int(input())
count=0
for i in range(1,11):
 if(n%i==0):
  count=count+1
if(count==2):
  print("prime")
else:
  print("not prime")
```

```python
#multiplication table
n=int(input())
for i in range(1,11):
  print(f"{n}*{i}={n*i}")
```
