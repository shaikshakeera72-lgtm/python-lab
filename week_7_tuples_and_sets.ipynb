```python

#OPERATIONS ON SETS
#1.Union (|)
#2.Intersection (&)
#3.Difference (-)
#4.Symmetric Difference (^)
a={1,2,3}
b={"Sri",4,5,6.0,3,2}
print(a&b) #intersection
print(a|b) #union
print(a-b) #difference
print(b-a) #difference
print(a^b) #Symmetric difference
print(a)
print(b)
```

```python

#To change the original set
c={11,22,33}
a={1,2,3,33}
print(a.union(c))
print(a.difference(c))
print(a.intersection(c))
print(a.intersection_update(c))
print(a) # a is intersection set contains only 33
```

```python
#number of elements repeated

L1=[i for i in input().split()]
L2=[]
for i in L1:
  if L1.count(i)>1:
    L2.append(i)
L3=set(L2)
print(len(L3))
```

```python
#Write a python program that prints common elements between two lists
L1=[i for  i in input().split()]
L2=[i for i in input().split()]
p=set(L1)
q=set(L2)
print(p & q)
```

```python
#Write a python program that find missing number between 1 and n using a set
n=int(input())
L1=list(map(int,input().split()))
L2=[]
for i in range(1,n+1):
  L2.append(i)
p=set(L1)
q=set(L2)
print(q-p)
```

```python
#Write a python program that prints only unique elements in a list

L1=[i for i in input().split()]
L2=[]
for i in L1:
  if L1.count(i)==1:
    L2.append(i)
print(L2)
```
