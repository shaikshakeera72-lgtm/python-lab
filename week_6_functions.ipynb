```python
#To check whether given number is palindrome or not using functions
def palindrome(num):
  original = num
  rev= 0
  while (num > 0):
    rem = num % 10
    rev = rev * 10 +rem
    num = num // 10
  if (original == rev):
    print("Palindrome")
  else:
    print("Not a palindrome")
n=int(input())
palindrome(n)


```

```python
#To reverse a string
def reversestring(x):
  print(x[::-1])
x=(input())
reversestring(x)
```

```python
def reverse_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
x=(input())
print(reverse_str(x))
```

```python
#function that return multipple
def multiple_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"sum is {res_sum},count is {res_count},the average is {res_avg}")
```
