```python
#palindrome or not
def number():
  input=input("enter a number:")
  n= input.strip()
  reversed_input=n[::-1]
  if n==reversed_input:
    print("yes,it is a palindrome")
  else:
      print("not a palindrome")
```

```python
#reverse a string using function
def reverse_string(s):
  return s[::-1]
s=input("enter a string:")
print(reverse_string(s))
```
